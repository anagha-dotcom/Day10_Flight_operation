## 7. Data Transformation

```python
df['Delay_Category'] = df['Delay_Minutes'].apply(
    lambda x: 'On Time' if x == 0
    else 'Minor Delay' if x <= 30
    else 'Major Delay'
)
```

```python
df[['Delay_Minutes', 'Delay_Category']].head(10)
```

```python
df['Seat_Utilization'] = (
    df['Passengers'] / df['Seat_Capacity'] * 100
)
```

```python
df[['Passengers', 'Seat_Capacity', 'Seat_Utilization']].head(10)
```

```python
df['Seat_Utilization'] = df['Seat_Utilization'].round(2)
```

## 8. DateTime Analysis

```python
df['Flight_Date'] = pd.to_datetime(df['Flight_Date'])
```

```python
df['Flight_Year'] = df['Flight_Date'].dt.year
```

```python
df['Flight_Month'] = df['Flight_Date'].dt.month
```

```python
df['Flight_Day'] = df['Flight_Date'].dt.day
```

```python
df['Flight_Day_Name'] = df['Flight_Date'].dt.day_name()
```

```python
df[['Flight_Date', 'Flight_Year', 'Flight_Month', 'Flight_Day', 'Flight_Day_Name']].head(10)
```

```python
df['Flight_Month'].value_counts().sort_index()
```

```python
df.groupby('Flight_Month').size().sort_values(ascending=False)
```

```python
df['Flight_Day_Name'].value_counts()
```

```python
# Flight status analysis
print("Flights by Status:")
print(df['Flight_Status'].value_counts())

# Average delay by weather condition
print("\nAverage Delay by Weather:")
print(df.groupby('Weather')['Delay_Minutes'].mean().sort_values(ascending=False))

# Passengers by travel class
print("\nPassengers by Travel Class:")
print(df.groupby('Travel_Class')['Passengers'].sum().sort_values(ascending=False))

# Average satisfaction by booking channel
print("\nAverage Satisfaction by Booking Channel:")
print(df.groupby('Booking_Channel')['Passenger_Satisfaction'].mean().sort_values(ascending=False))

# Most common destination
print("\nMost Common Destinations:")
print(df['Destination'].value_counts().head(10))
```

## 10. Key Observations

1. Most flights were **On Time (114)**, while **59 flights were Delayed** and **7 flights were Cancelled**.

2. Flights during **Rainy weather conditions** had the highest average delay of approximately **22.22 minutes**.

3. **Economy class** had the highest number of passengers with **16,054 passengers**, followed by Premium Economy and Business class.

4. Passengers who booked through an **Online Travel Portal** had the highest average satisfaction score of **3.74**.

5. **Delhi** was the most common destination with **51 flights**, followed by Mumbai and Bengaluru.

6. Among the weather conditions, **Fog** had the lowest average delay of approximately **15.79 minutes**.

7. The number of cancelled flights was relatively low, with only **7 cancelled flights** in the dataset.

8. The analysis shows that weather conditions, travel class, booking channel, and destination are useful factors for understanding flight operations and passenger experience.
